```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Configuration de la figure
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection="3d")

# Référentiel de l'observateur : Z = zénith, X = Nord, Y = Est
ax.set_xlabel("Nord (X)", fontsize=12)
ax.set_ylabel("Est (Y)", fontsize=12)
ax.set_zlabel("Zénith (Z)", fontsize=12)
ax.set_xlim([-1, 1])
ax.set_ylim([-1, 1])
ax.set_zlim([0, 1])

# Axe optique du télescope (pointe vers ζ_tel = 30°, A_tel = 0°)
zeta_tel = np.radians(30)  # Angle zénithal du télescope
A_tel = np.radians(0)  # Azimut du télescope (Nord)
x_tel = np.sin(zeta_tel) * np.cos(A_tel)
y_tel = np.sin(zeta_tel) * np.sin(A_tel)
z_tel = np.cos(zeta_tel)
ax.quiver(0, 0, 0, x_tel, y_tel, z_tel, color="black", linewidth=3, label="Axe optique du télescope")

# Direction de l'objet céleste (ζ = 45°, A = 30°)
zeta_obj = np.radians(45)  # Angle zénithal de l'objet
A_obj = np.radians(30)  # Azimut de l'objet
x_obj = np.sin(zeta_obj) * np.cos(A_obj)
y_obj = np.sin(zeta_obj) * np.sin(A_obj)
z_obj = np.cos(zeta_obj)
ax.quiver(
    0, 0, 0, x_obj, y_obj, z_obj, color="gray", linewidth=2, linestyle="--", label="Direction de l'objet"
)

# Calcul de l'angle parallactique θ
cos_theta = np.sin(zeta_obj) * np.sin(zeta_tel) + np.cos(zeta_obj) * np.cos(zeta_tel) * np.cos(A_obj - A_tel)
theta = np.degrees(np.arccos(cos_theta))

# Affichage de θ
ax.text(0.1, 0.1, 0.5, f"θ = {theta:.1f}° (angle parallactique)", fontsize=12, color="darkred")

# Simulation de la réfraction atmosphérique pour deux longueurs d'onde (bleu et rouge)
# On suppose une masse d'air X = 1 / cos(ζ_obj)
X = 1 / np.cos(zeta_obj)
# Coefficient de réfraction approximatif (plus fort pour le bleu)
chi_blue = 0.1 * X  # Exagéré pour la visualisation
chi_red = 0.05 * X  # Moins fort pour le rouge

# Rayons incidents (avant réfraction)
rayon_blue = np.array([x_obj, y_obj, z_obj])
rayon_red = np.array([x_obj, y_obj, z_obj])

# Après réfraction : déviation vers le zénith (simplifiée)
# On applique une déviation dans le plan vertical (vers le zénith)
rayon_blue_refracted = rayon_blue + np.array([0, 0, chi_blue])
rayon_red_refracted = rayon_red + np.array([0, 0, chi_red])

# Normalisation pour l'affichage
rayon_blue_refracted = rayon_blue_refracted / np.linalg.norm(rayon_blue_refracted)
rayon_red_refracted = rayon_red_refracted / np.linalg.norm(rayon_red_refracted)

# Tracé des rayons réfractés
ax.quiver(
    0,
    0,
    0,
    rayon_blue_refracted[0],
    rayon_blue_refracted[1],
    rayon_blue_refracted[2],
    color="blue",
    linewidth=2,
    label="Rayon bleu (λ court, χ fort)",
)
ax.quiver(
    0,
    0,
    0,
    rayon_red_refracted[0],
    rayon_red_refracted[1],
    rayon_red_refracted[2],
    color="red",
    linewidth=2,
    label="Rayon rouge (λ long, χ faible)",
)

# Plan focal du télescope (simplifié : à z = 0.9)
z_focal = 0.9
ax.plot_surface([-1, 1], [-1, 1], [z_focal, z_focal], color="lightgray", alpha=0.3, label="Plan focal")

# Projection des rayons réfractés sur le plan focal
# On calcule l'intersection avec le plan z = z_focal
t_blue = (z_focal - 0) / rayon_blue_refracted[2]
t_red = (z_focal - 0) / rayon_red_refracted[2]

x_blue_focal = rayon_blue_refracted[0] * t_blue
y_blue_focal = rayon_blue_refracted[1] * t_blue
x_red_focal = rayon_red_refracted[0] * t_red
y_red_focal = rayon_red_refracted[1] * t_red

# Tracé des points d'impact sur le plan focal
ax.scatter(x_blue_focal, y_blue_focal, z_focal, color="blue", s=100, label="Impact bleu (décalé)")
ax.scatter(x_red_focal, y_red_focal, z_focal, color="red", s=100, label="Impact rouge (décalé)")

# Ligne reliant les impacts (dipôle)
ax.plot(
    [x_blue_focal, x_red_focal],
    [y_blue_focal, y_red_focal],
    [z_focal, z_focal],
    color="purple",
    linewidth=3,
    label="Dipôle (ADR)",
)

# Légende et titre
ax.set_title(
    "Angle parallactique (θ) et effet de l'ADR (Atmospheric Differential Refraction)\n"
    "Référentiel : Z = Zénith, X = Nord, Y = Est",
    fontsize=14,
    pad=20,
)
ax.legend(loc="upper right", fontsize=10)

# Affichage
plt.tight_layout()
plt.show()
```
